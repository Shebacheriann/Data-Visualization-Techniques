```python
import pandas as pd

df = pd.read_csv(r"C:\Users\HP\Downloads\archive\automobile_dataset.csv")
print("===== FULL DATA =====")
print(df)

print("\n===== HEAD (First 5 Rows) =====")
print(df.head())

print("\n===== INFO =====")
print(df.info())

print("\n===== DESCRIBE (Summary Statistics) =====")
print(df.describe())

# ============================================================
# 2. BASIC INFO
# ============================================================
print("\n===== SHAPE =====")
print(df.shape)

print("\n===== DATA TYPES =====")
print(df.dtypes)

# ============================================================
# 3. FILTER COLUMNS
# ============================================================
print("\n===== FILTER COLUMNS (Make, Year, Selling_Price) =====")
print(df[["Make", "Year", "Selling_Price"]])

print("\n===== FILTER COLUMNS (Fuel_Type, Body_Type, Mileage) =====")
print(df[["Fuel_Type", "Body_Type", "Mileage"]])

# ============================================================
# 4. FILTER ROWS
# ============================================================
print("\n===== Year > 2015 =====")
print(df[df["Year"] > 2015])

print("\n===== Fuel_Type = Petrol =====")
print(df[df["Fuel_Type"] == "Petrol"])

print("\n===== Transmission = Automatic =====")
print(df[df["Transmission"] == "Automatic"])

print("\n===== Body_Type = SUV =====")
print(df[df["Body_Type"] == "SUV"])

print("\n===== Accident_History = 1 (Yes) =====")
print(df[df["Accident_History"] == 1])

# Multiple Conditions
print("\n===== Year > 2015 AND Transmission = Automatic =====")
print(df[(df["Year"] > 2015) & (df["Transmission"] == "Automatic")])

print("\n===== Mileage > 50000 AND Fuel_Type = Diesel =====")
print(df[(df["Mileage"] > 50000) & (df["Fuel_Type"] == "Diesel")])
```

```python
# 5. SORTING
# ============================================================
print("\n===== SORT BY Year (Ascending) =====")
print(df.sort_values(by="Year"))

print("\n===== SORT BY Year (Descending) =====")
print(df.sort_values(by="Year", ascending=False))

print("\n===== SORT BY Selling_Price (Descending) =====")
print(df.sort_values(by="Selling_Price", ascending=False))

print("\n===== SORT BY Make AND Year =====")
print(df.sort_values(by=["Make", "Year"]))

print("\n===== SORT BY Body_Type AND Mileage =====")
print(df.sort_values(by=["Body_Type", "Mileage"]))

# ============================================================
# 6. STATISTICS
# ============================================================
print("\n===== STATISTICS ON Selling_Price =====")
print("Mean Selling_Price   =", df["Selling_Price"].mean())
print("Median Selling_Price =", df["Selling_Price"].median())
print("Mode Selling_Price   =")
print(df["Selling_Price"].mode())
print("Std Dev Selling_Price =", df["Selling_Price"].std())

print("\n===== STATISTICS ON Mileage =====")
print("Mean Mileage   =", df["Mileage"].mean())
print("Median Mileage =", df["Mileage"].median())
print("Std Dev Mileage =", df["Mileage"].std())

print("\n===== STATISTICS ON Year =====")
print("Mean Year   =", df["Year"].mean())
print("Median Year =", df["Year"].median())

# ============================================================
# 7. GROUP BY
# ============================================================
print("\n===== Average Selling_Price by Fuel_Type =====")
fuel_avg = df.groupby("Fuel_Type")["Selling_Price"].mean()
print(fuel_avg)

print("\n===== Average Selling_Price by Body_Type =====")
body_avg = df.groupby("Body_Type")["Selling_Price"].mean()
print(body_avg)

print("\n===== Average Selling_Price by Make =====")
make_avg = df.groupby("Make")["Selling_Price"].mean().sort_values(ascending=False)
print(make_avg)

print("\n===== Average Mileage by Transmission =====")
print(df.groupby("Transmission")["Mileage"].mean())

# ============================================================
# 8. CORRELATION
# ============================================================
print("\n===== CORRELATION (Year vs Selling_Price) =====")
print(df["Year"].corr(df["Selling_Price"]))

print("\n===== CORRELATION (Mileage vs Selling_Price) =====")
print(df["Mileage"].corr(df["Selling_Price"]))

print("\n===== CORRELATION (Engine_Size vs Horsepower) =====")
print(df["Engine_Size"].corr(df["Horsepower"]))

print("\n===== CORRELATION (Horsepower vs Selling_Price) =====")
print(df["Horsepower"].corr(df["Selling_Price"]))

print("\n===== FULL PEARSON CORRELATION (Numeric Columns) =====")
print(df.corr(numeric_only=True, method='pearson'))

print("\n===== FULL SPEARMAN CORRELATION (Numeric Columns) =====")
print(df.corr(numeric_only=True, method='spearman'))

print("\n===== Correlation Matrix (selected columns) =====")
corr = df[["Year", "Mileage", "Engine_Size", "Horsepower", "Selling_Price"]].corr()
print(corr)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))
plt.scatter(df["Mileage"], df["Selling_Price"], alpha=0.5)
plt.title("Mileage vs Selling Price")
plt.xlabel("Mileage")
plt.ylabel("Selling Price")
plt.grid(True)
plt.tight_layout()
plt.show()

# Scatter Plot 2: Year vs Selling_Price
plt.figure(figsize=(7, 5))
plt.scatter(df["Year"], df["Selling_Price"], alpha=0.5, color="green")
plt.title("Year vs Selling Price")
plt.xlabel("Year")
plt.ylabel("Selling Price")
plt.grid(True)
plt.tight_layout()
plt.show()

# Bar Plot 1: Average Selling Price by Fuel_Type
fuel_avg.plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Average Selling Price by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average Selling Price")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Bar Plot 2: Average Selling Price by Body_Type
body_avg.plot(kind="bar", color="orange", edgecolor="black")
plt.title("Average Selling Price by Body Type")
plt.xlabel("Body Type")
plt.ylabel("Average Selling Price")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
