##### import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\HP\Downloads\archive\automobile_dataset.csv")

# ============================================================
# 1. Line Plot - Year vs Selling Price
# ============================================================
plt.figure(figsize=(8,5))
plt.plot(df["Year"], df["Selling_Price"], alpha=0.6)
plt.title("Line Plot: Year vs Selling Price")
plt.xlabel("Year")
plt.ylabel("Selling Price")
plt.grid(True)
plt.tight_layout()
plt.show()

# ============================================================
# 2. Box Plot - Selling Price
# ============================================================
plt.figure(figsize=(6,5))
plt.boxplot(df["Selling_Price"].dropna())
plt.title("Selling Price Box Plot")
plt.ylabel("Selling Price")
plt.tight_layout()
plt.show()

# ============================================================
# 3. Histogram - Mileage Distribution
# ============================================================
plt.figure(figsize=(8,5))
plt.hist(df["Mileage"].dropna(), bins=20, edgecolor="black", color="skyblue")
plt.title("Mileage Distribution")
plt.xlabel("Mileage")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# ============================================================
# 4. Violin Plot - Selling Price
# ============================================================
plt.figure(figsize=(6,5))
plt.violinplot(df["Selling_Price"].dropna())
plt.title("Selling Price Violin Plot")
plt.ylabel("Selling Price")
plt.tight_layout()
plt.show()

# ============================================================
# 5. Line Plot - Selling Price Trend (by index)
# ============================================================
plt.figure(figsize=(10,5))
plt.plot(df.index, df["Selling_Price"], color="red", alpha=0.7)
plt.title("Selling Price Trend")
plt.xlabel("Record Index")
plt.ylabel("Selling Price")
plt.grid(True)
plt.tight_layout()
plt.show()

# ============================================================
# 6. Stack Plot - Engine Size & Horsepower
# ============================================================
plt.figure(figsize=(10,5))
plt.stackplot(
    df.index,
    df["Engine_Size"].fillna(0),
    df["Horsepower"].fillna(0),
    labels=["Engine Size", "Horsepower"]
)
plt.title("Engine Size and Horsepower")
plt.xlabel("Record Index")
plt.ylabel("Values")
plt.legend(loc="upper left")
plt.tight_layout()
plt.show()

# ============================================================
# 7. Line Plot - Mileage by Index
# ============================================================
plt.figure(figsize=(12,5))
plt.plot(df.index, df["Mileage"], color="green", alpha=0.7)
plt.title("Mileage by Record Index")
plt.xlabel("Record Index")
plt.ylabel("Mileage")
plt.tight_layout()
plt.show()

# ============================================================
# 8. Pie Chart - Fuel Type Distribution
# ============================================================
fuel = df["Fuel_Type"].value_counts()

plt.figure(figsize=(7,7))
plt.pie(fuel,
        labels=fuel.index,
        autopct="%1.1f%%",
        startangle=90)
plt.title("Fuel Type Distribution")
plt.tight_layout()
plt.show()

# ============================================================
# 9. Subplots (2x2)
# ============================================================
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

# Line Plot
ax[0,0].plot(df.index, df["Selling_Price"], color="purple", alpha=0.6)
ax[0,0].set_title("Line Plot - Selling Price")

# Histogram
ax[0,1].hist(df["Year"].dropna(), bins=15, edgecolor="black", color="orange")
ax[0,1].set_title("Histogram - Year")

# Box Plot
ax[1,0].boxplot(df["Mileage"].dropna())
ax[1,0].set_title("Box Plot - Mileage")

# Scatter Plot
ax[1,1].scatter(df["Engine_Size"], df["Horsepower"], alpha=0.5)
ax[1,1].set_title("Scatter - Engine Size vs Horsepower")
ax[1,1].set_xlabel("Engine Size")
ax[1,1].set_ylabel("Horsepower")

plt.tight_layout()
plt.show()

# ============================================================
# 10. Seaborn Scatter Plot
# ============================================================
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Mileage", y="Selling_Price", alpha=0.6)
plt.title("Mileage vs Selling Price (Seaborn)")
plt.tight_layout()
plt.show()
